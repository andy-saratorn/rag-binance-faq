```python
import os
import json
from langchain_core.documents import Document
from langchain_community.retrievers import BM25Retriever
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_chroma import Chroma
from dotenv import load_dotenv
from pydantic import BaseModel, Field
from typing import List


load_dotenv()
```

```python
with open('./json/test_questions.json', 'r') as f:
    # Parsing the JSON file into a Python dictionary
    test_questions = json.load(f)
```

# Multi Queries

```python
class Queries(BaseModel):
    queries: List[str] = Field(description="List of search queries")
```

```python
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
llm_with_tool = llm.with_structured_output(Queries)
```

# Chroma DB Retriever

```python
# 1. Setup the same embedding model and directory
persist_directory = "db/Chroma_db"
embedding_model = OpenAIEmbeddings(model="text-embedding-3-small")

# 2. Load the existing database from disk
vectorstore = Chroma(
    persist_directory=persist_directory, 
    embedding_function=embedding_model
)

vector_retriever = vectorstore.as_retriever(search_kwargs={'k':3})
```

# Final Answer

```python
test_question = test_questions[0]
```

```python
for question in test_question['questions'][1:2]:
    # Query Variations
    prompt = f"""
        Generate 3 diverse search queries for: {question}.
        Each query would try to approach the provided question from a different angle.
    """
    response = llm_with_tool.invoke(prompt)
    query_variations = response.queries
    # Relevant Docs
    relevant_docs = []
    for query in query_variations:
        relevant_docs_i = vector_retriever.invoke(query)
        relevant_docs += relevant_docs_i
    # Final answer    
    messages = [
        SystemMessage(content="Answer the question using imformation from the given documents only. If there're no relevant information, just say there're no information"),
        *[doc.page_content for doc in relevant_docs],    
        HumanMessage(content=f'The question: {question}')
    ]
    
    final_answer_response = llm.invoke(messages)
    final_answer = final_answer_response.content.strip()
    print(f'\nQuestion: {question}')
    print(f'\nAnswer for: {final_answer}')
```
